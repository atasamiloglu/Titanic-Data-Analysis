```python
childs = df.query("age < 12")

len(childs)                        
childs["survived"].sum()            
childs["survived"].mean()        

# Survival rates of children under 12 years old
```

```python
df.query("embark_town == 'Cherbourg' and sex == 'male'" )["fare"].mean()

# Average ticket prices of male passengers who embarked at Cherbourg
```

```python
alone_group = df[df["alone"] == True]
alone_group['survived'].mean()

# Survival rates of solo passengers
```

```python
family_group = df[df["alone"] == False]
family_group["survived"].mean()

# Survival rates of passengers traveling with family
```

```python
# Passengers traveling with family had a higher survival rate than passengers traveling alone.
```

```python
df.groupby("sex")["survived"].mean()

# Survival rates according to the gender of the passengers.
```

```python
df.groupby("pclass")[["age", "fare"]].agg(["mean","std","count"])

# Average age and fare by passenger class
```

```python
df.groupby(["sex", "pclass"])["survived"].agg(["mean", "count"])
```

```python
df.groupby(["sex","pclass"])["survived"].mean()

# Survival rate by sex and passenger class
```

```python
sns.set_style("whitegrid")
g = sns.displot(x="age", data=df, hue="sex", col="sex")
g.figure.suptitle("Age Distributor of Titanic Passengers")
```

```python
sns.boxplot(data=df, x="pclass", y="fare")
plt.title("Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()
```

```python
# Interesting observation:
# Some 3rd class passengers paid more than the median fare of 1st class,
# yet they were still classified as 3rd class.
# Perhaps they had a lower chance of survival because of their class,
# but this is only a possibility, not something this plot can prove.
```

```python
sns.countplot(data=df, x="sex", hue="survived")
plt.title("Survival Counts by Sex")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.scatterplot(data=df, x="age", y="fare", hue="survived")
plt.title("Age vs Fare by Survival")
```

```python
# Key observation:
# Passengers who paid higher fares generally survived,
# while most of the passengers who paid the lowest fares did not.
# Note: this shows a relationship, not a cause. Fare is tied to passenger class,
# so it probably reflects class (and sex) rather than the fare itself.
```
