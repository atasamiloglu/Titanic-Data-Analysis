```python
childs = df.query("age < 12")

len(childs)                        
childs["survived"].sum()            
childs["survived"].mean()        

# Survival rates of children under 12 years old
```

```python
df.query("embark_town == 'Cherbourg' and sex == 'male'" )["fare"].mean()

# Average ticket prices of male passengers who embarked at Cherbourg
```

```python
alone_group = df[df["alone"] == True]
alone_group['survived'].mean()

# Survival rates of solo passengers
```

```python
family_group = df[df["alone"] == False]
family_group["survived"].mean()

# Survival rates of passengers traveling with family
```

```python
# Passengers traveling with family had a higher survival rate than passengers traveling alone.
```
